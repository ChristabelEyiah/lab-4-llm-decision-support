````python
import os

# --- Google Colab (Secrets panel) ---
from google.colab import userdata
API_KEY = userdata.get("GROQ_API_KEY")

# OpenAI-compatible client
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1",
)

MODEL = "llama-3.3-70b-versatile"

print("Client ready.")
````

````python
def ask_llm(user_prompt, system_prompt="You are a helpful assistant.",
            temperature=0.7, max_tokens=500):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response


# Test the function
response = ask_llm("What is artificial intelligence?")

print("Answer:")
print(response.choices[0].message.content)

print("\nToken usage:")
print(response.usage)
````

**Student Reasoning - Anatomy of A Call**

1. What is the difference between the system and user roles?

The system role gives the LLM general instructions about how it should behave or respond. It sets the context, rules, or personality for the conversation. The user role contains the actual question or task that the user wants the LLM to answer.

2. What is a token, roughly? Why do API providers bill per token rather than per request?

A token is a small piece of text that an LLM processes. It can be a whole word, part of a word, punctuation, or another small unit of text.

API providers bill based on tokens because the amount of computation required depends largely on how much text the model has to process and generate, not simply on the number of API requests. A short request might use only a few tokens, while a long conversation or response could use thousands. Charging per token therefore more accurately reflects the resources used by the model.



````python
question = "Suggest a name for a savings product for market traders in Accra."

# Temperature = 0.0
print("TEMPERATURE 0.0")

for i in range(5):
    response = ask_llm(
        question,
        temperature=0.0,
        max_tokens=100
    )
    print(f"\nAnswer {i+1}:")
    print(response.choices[0].message.content)


# Temperature = 1.2
print("\n\nTEMPERATURE 1.2 ")

for i in range(5):
    response = ask_llm(
        question,
        temperature=1.2,
        max_tokens=100
    )
    print(f"\nAnswer {i+1}:")
    print(response.choices[0].message.content)
````

**Student Reasoning - Temperature**
At temperature 0.0, the responses were generally more consistent and similar to each other. The model tended to give the same or very similar suggestions because there is less randomness in the generation process.

At temperature 1.2, the responses were more varied and creative. The model produced a wider range of possible names, but the results were also less predictable.

For the loan decision-support system, a low temperature such as 0.0 or around 0.2–0.3 is more appropriate. Loan decisions should be consistent, predictable, and based on the information provided rather than creative or random responses. A low temperature reduces unnecessary variation and makes the system more reliable when providing explanations or recommendations.

````python
# SUMMARY_PROMPT_V2

SUMMARY_SYSTEM_V2 = """You are an assistant to a microfinance loan officer.
Summarize loan application letters into a brief factual summary.
Be neutral and objective. Do not invent, assume, or infer information that is
not explicitly stated in the letter. Include the applicant, requested loan
amount, purpose of the loan, business/financial situation, and repayment or
security information when available. Keep the summary to 3-4 sentences."""

SUMMARY_PROMPT_V2 = """Summarize this loan application:

{letter_text}"""
````

**Student Reasoning — Summarization Prompts**

1. What concrete problems did V1's output have that V2 fixed? Quote examples.

V1 was too general because the prompt only said “Summarize this”, so the model was not given specific requirements about what information to prioritize. For example, V1 may have produced a general description of the applicant's situation while leaving out important loan details such as the requested amount, purpose, repayment plan, or lack of collateral. V2 fixed this by explicitly instructing the model to include the applicant, loan amount, purpose, financial situation, and repayment/security information.

2. Why is “no invented details” an essential instruction in this application? What is this failure mode called in the LLM literature?

“No invented details” is essential because this is a loan decision-support system, where inaccurate information could influence whether someone receives a loan. The LLM must only use facts stated in the application and should not assume that an applicant has income, collateral, repayment ability, or other qualifications that were not provided. Inventing information could therefore lead to unfair or financially harmful decisions.

This failure mode is called hallucination in the LLM literature. A hallucination occurs when a model generates information that is unsupported by, or inconsistent with, the provided source.

````python
EXTRACT_PROMPT = """
Extract the required loan application information from the letter below.

Return ONLY a valid JSON object. Do not include markdown, explanations,
comments, or any text outside the JSON object.

The JSON object MUST contain EXACTLY these keys:

{{
  "applicant_name": "string",
  "amount_ghs": "number",
  "purpose": "string",
  "monthly_profit_ghs": "number or null",
  "has_collateral_or_guarantor": "boolean",
  "repayment_months": "number or null"
}}

Rules:
- applicant_name must be the applicant's name as stated in the letter.
- amount_ghs must be the requested loan amount in Ghana cedis.
- purpose must describe what the applicant says the loan will be used for.
- monthly_profit_ghs must be the stated monthly profit, or null if it is not stated.
- has_collateral_or_guarantor must be true if the letter states that the applicant
  has collateral or a guarantor, and false if neither is stated.
- repayment_months must be the stated repayment period in months, or null if it
  is not stated.
- If a field is not stated in the letter, use null. Do not guess.
- Use numbers for numerical values, not strings.
- Use true or false for the boolean field.

Worked example:

Letter:
"My name is Ama Osei. I run a small bakery in Accra. I am requesting
GHS 6,000 to buy an oven. My business makes GHS 1,200 profit per month.
I will repay the loan over 10 months. My brother will guarantee the loan."

Correct JSON:
{{
  "applicant_name": "Ama Osei",
  "amount_ghs": 6000,
  "purpose": "buy an oven",
  "monthly_profit_ghs": 1200,
  "has_collateral_or_guarantor": true,
  "repayment_months": 10
}}

Now extract the information from this loan application:

{letter_text}
"""
````

````python
import json

def extract_fields(letter_text):
    prompt = EXTRACT_PROMPT.format(letter_text=letter_text)

    try:
        response = ask_llm(
            prompt,
            temperature=0,
            max_tokens=300
        )

        result = response.choices[0].message.content.strip()

        # Remove markdown JSON fences if the model adds them
        if result.startswith("```json"):
            result = result[len("```json"):].strip()

        if result.startswith("```"):
            result = result[len("```"):].strip()

        if result.endswith("```"):
            result = result[:-3].strip()

        # Convert JSON string into Python dictionary
        data = json.loads(result)

        return data

    except json.JSONDecodeError:
        print("Warning: Model returned invalid JSON.")
        return None

    except Exception as e:
        print(f"Warning: Extraction failed: {e}")
        return None
````

**Student Reasoning — Decision Support**

1. Compare the briefs for L003 (strong application) and L006 (weak application). Did the system identify the right strengths and red flags in each?

Yes. The system identified important differences between the two applications. L003 is a stronger application because Efua Darko has an established registered business, employs three apprentices, earns an average monthly profit of GHS 2,800, has a GHS 5,000 fixed deposit that can be pledged, and provides a specific repayment plan of GHS 1,100 monthly for 15 months. These details provide evidence of an operating business, income, and security for the loan.

In contrast, L006 has several red flags. Kofi is requesting GHS 50,000 to start three different businesses that have not yet been established. He does not provide an existing business income or monthly profit, has no collateral, and gives a vague repayment plan based on the businesses "booming." The system therefore correctly identified L003's stronger financial evidence and security while highlighting L006's lack of established income, collateral, and concrete repayment capacity.

2. Why did we forbid the model from outputting "approve"/"reject"? Give one practical and one ethical reason.

Practically, the LLM may make mistakes or misunderstand information in the application, so allowing it to directly approve or reject loans could lead to incorrect financial decisions. The system should instead provide factual information and risk indicators that a qualified human can review.

Ethically, lending decisions can have a significant impact on people's lives and financial opportunities. Allowing an AI to make the final decision could create unfair or biased outcomes without sufficient human oversight or accountability. Keeping the model as a decision-support tool ensures that a human remains responsible for the final lending decision.

````python
fields = [
    "applicant_name",
    "amount_ghs",
    "purpose",
    "monthly_profit_ghs",
    "has_collateral_or_guarantor",
    "repayment_months"
]

# Make letter_id the index for easy lookup
df_eval = df.set_index("letter_id")

rows = []

for field in fields:
    row = {"field": field}
    correct = 0

    for letter_id in GOLD.keys():
        predicted = df_eval.loc[letter_id, field]
        gold_value = GOLD[letter_id][field]

        # Case-insensitive comparison for names
        if field == "applicant_name":
            match = (
                str(predicted).strip().lower()
                == str(gold_value).strip().lower()
            )

        # Exact comparison for numbers, booleans, and null
        else:
            match = predicted == gold_value

        row[letter_id] = "✓" if match else "✗"

        if match:
            correct += 1

    row["accuracy"] = correct / len(GOLD)
    rows.append(row)

accuracy_df = pd.DataFrame(rows)

display(accuracy_df)
````

````python
import json

def extract_fields(letter_text, temperature=0):
    prompt = EXTRACT_PROMPT.format(letter_text=letter_text)

    try:
        response = ask_llm(
            prompt,
            temperature=temperature,
            max_tokens=300
        )

        result = response.choices[0].message.content.strip()

        # Remove markdown JSON fences if the model adds them
        if result.startswith("```json"):
            result = result[len("```json"):].strip()

        if result.startswith("```"):
            result = result[len("```"):].strip()

        if result.endswith("```"):
            result = result[:-3].strip()

        data = json.loads(result)

        return data

    except json.JSONDecodeError:
        print("Warning: Model returned invalid JSON.")
        return None

    except Exception as e:
        print(f"Warning: Extraction failed: {e}")
        return None
````

````python
import json

letter = LETTERS["L004"]

results_temp_0 = []
results_temp_1 = []

# Temperature 0.0 — 5 runs
for i in range(5):
    result = extract_fields(letter, temperature=0.0)
    results_temp_0.append(result)

# Temperature 1.0 — 5 runs
for i in range(5):
    result = extract_fields(letter, temperature=1.0)
    results_temp_1.append(result)
````

````python
def reliability_report(results):
    # Keep only valid JSON results
    valid_results = [r for r in results if r is not None]

    # Convert dictionaries to standardized JSON strings
    unique_results = set(
        json.dumps(r, sort_keys=True)
        for r in valid_results
    )

    return {
        "total_runs": len(results),
        "valid_json": len(valid_results),
        "unique_valid_results": len(unique_results),
        "identical_across_all_valid_runs": len(unique_results) == 1
    }


report_0 = reliability_report(results_temp_0)
report_1 = reliability_report(results_temp_1)

print("Temperature 0.0:")
print(report_0)

print("\nTemperature 1.0:")
print(report_1)
````
